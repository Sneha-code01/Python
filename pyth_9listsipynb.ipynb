```python
list = [1,2,3]
#append adds one element at the end
list.append(4)
print(list)
#sort is use to arrange list in ascending orderr
list.sort()
list.sort(reverse= True)#sort in descending order
print(list)
list.insert(0, 5)#i,v ie i for index jispe v value jayega
print(list)
```
