```python
import pandas as pd
import numpy as np

csv1 = pd.read_csv("/content/ipl-matches.csv")
print(csv1)
print(type(csv1))
#single column fetch
print(csv1['Player_of_Match'])
print(csv1[['Team1','Team2','WinningTeam']])


```

```python
std_dict = {
    'name': ['Sneha','Shruti','Abadh','Kishore','Yadav','Choudhary'],
    'IQ' : [120,110,80,90,100,70],
    'Marks' : [90,85,100,70,60,80],
    'LPA' : [10,12,15,8,9,13]
}
#fetching row1 iloc: searches using index positions and loc: searches using index labels
dft1 = pd.DataFrame(std_dict)
print(dft1)
dft1.set_index('name',inplace=True)
print(dft1)
#single row
print("single row required:\n",dft1.iloc[0])
#multiple row
print("multiple row together:\n", dft1.iloc[2:4])
print("rows through label:\n",dft1.loc['Abadh':'Choudhary'])#loc takes also the highest value of range
#fancy indexing
print("taking out rows without pattern:\n",dft1.iloc[[0,4,2]])
```

```python
#filtering data
final = csv1['MatchNumber'] == 'Final'
print(final)
csvfinal = csv1[final]
print(csvfinal[['Season','WinningTeam']])
```

```python
#checking total superover

csv_superover = csv1[csv1['SuperOver']== 'Y']
print(csv_superover['SuperOver'])
print("total matches having super over:",csv_superover['SuperOver'].shape[0])
```

```python
#csk won in kolkata
csk_kolkata = csv1[csv1['City']=='Kolkata']
print(csk_kolkata[['City','WinningTeam']])
csk = csk_kolkata[['City','WinningTeam']]
csk_win = csk[csk['WinningTeam']=='Chennai Super Kings']
print(csk_win)
print("total matches csk won in kolkata ",csk_win.shape[0])


#short way
csk_win2 = csv1[(csv1['City'] == 'Kolkata') & (csv1['WinningTeam']== 'Chennai Super Kings')]
print("short method using &\n",csk_win2[['City','WinningTeam']])
```

```python
#toss winner is match winner
toss_mtch =  csv1[csv1['TossWinner'] == csv1['WinningTeam']]
print(toss_mtch[['TossWinner','WinningTeam']])
print("total no of matches:\n",csv1.shape[0])
print("total number of matches won by toss winning team:\n",toss_mtch[['TossWinner','WinningTeam']].shape[0])
i = csv1.shape[0]
f = toss_mtch[['TossWinner','WinningTeam']].shape[0]
per = ((i-f)/i)*100
print("chances the toss winner is match winner %:\n",per)
```
