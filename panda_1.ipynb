```python
import numpy as np
import pandas as pd

#by using python list
num = [11,12,13,14,15]

print(num)
print(pd.Series(num))
#custom indexing in panda series
marks = [80,70,60,60]
subject = ["FM","AEM","DOM","IE"]
print(pd.Series(marks,index=subject,name = "Sneha ke marks"))

#series using dictionary
mark1 = {"FM":75,"AEM":67,"DOM":54,"IE":58}
print(pd.Series(mark1,name="Anjali ke marks"))

#series attributes 1.name 2.is_unique 3.index 4.values

```
