```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#scatter plot category colouring
plt.figure(figsize=(15,7.5))
iris = pd.read_csv("/content/iris.csv")
print(iris.sample(6))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],alpha=1)
plt.colorbar()
```

```python
#annotation
batu = pd.read_csv("/content/batter (1).csv")
batu1 = batu.head(100).sample(25)
print(batu1)
plt.figure(figsize=(17,17))
plt.scatter(batu1['avg'],batu1['strike_rate'],s=batu1['runs']*5/15,c=batu1['runs'])
plt.colorbar()
plt.axhline(130,color='red')
plt.axvline(30,color='red')
#horizontal and verrtical line

for i in range(batu1.shape[0]):
  plt.text(batu1['avg'].values[i],batu1['strike_rate'].values[i],batu1['batter'].values[i])
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2)
batu2= batu.sample(4)
#ax is array and its 2d array so each positon will be defined with x and y and will be stored in figure

ax[0][0].scatter(batu2['avg'],batu2['runs'])
ax[0][1].bar(batu2['batter'],batu2['runs'])
ax[1][0].scatter(batu2['avg'],batu2['runs'])
ax[1][1].scatter(batu2['avg'],batu2['runs'])
```

```python
#3d scatter plot
fig = plt.figure()
ax= plt.subplot(projection='3d')
ax.scatter3D(batu['avg'],batu['strike_rate'],batu['runs'])
```

```python
x = np.linspace(-100,100,10)
y = np.linspace(-100,100,10)

xx,yy = np.meshgrid(x,y)

z = xx**2 + yy**2
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
```
