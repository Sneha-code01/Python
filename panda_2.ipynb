```python
import numpy as np
import pandas as pd

csv1 = pd.read_csv("/content/kohli_ipl.csv",index_col="match_no").squeeze('columns')
print(csv1)
#Series indexing
#1.integer indexing
print(csv1[215])
# panda itself provide an index along with your given index and will provide same result
# in string Series the +ve,-ve and 0 all indexing work
#in int series only +ve indexing work
str1 = pd.Series(["Sneha","Shruti","Abadh","Kishore","Pado"])
print(str1)
print("nagative slicing in string series\n",str1[0])#neg not working?

#series slicing
print("slice of vk run:\n",csv1[5:10])
#negative slicing
print("from last\n",csv1[-10::2])
```

```python
#editing Series
#using indexing
print(csv1)
csv1['216'] = 100
print(csv1)
#using slicing
csv1[1:5] = (50,80,100,20)
print(csv1)
```

```python
from typing_extensions import Dict
#series with python functionalities
print(len(csv1))
print(type(csv1))
print(max(csv1))
print(dir(csv1))
print(sorted(csv1))
#type conversion
print(type(list(csv1)))
print(type(dict(csv1)))
#membership operator
#in works in index only
#loops works in values
#arthematic operator
print("kitne run se vk ki century reh gyi\n",100-csv1)
#boolean indexing on series\
print("kitne mei 50 se kaam run\n",csv1[csv1 < 50].size)
```
