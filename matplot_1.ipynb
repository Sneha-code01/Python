```python
from matplotlib.lines import lineStyles
#matplotlib
#type of data: 1.numerical = age,number,marks 2.categorical = gender,degree,uni
#graph plotting depends on data type
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

run = pd.read_csv("/content/sharma-kohli.csv")
print(run)#cat=index num=run
plt.plot(run['index'],run['V Kohli'], color = 'red',linestyle='dashed', linewidth=2,marker='d',markersize=10,label='Virat')
plt.plot(run['index'],run['RG Sharma'],linewidth=2,marker='s',markersize=8,label='Rohit')
plt.xlabel('runs scored')
plt.ylabel('year of scoring')
plt.title('KOHLI V/S ROHIT')
plt.legend()
plt.grid()

```
