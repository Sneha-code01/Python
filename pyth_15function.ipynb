```python
def cal_avg(a,b,c):
  return ((a+b+c)/3)

def cal_sum(a,b):
  return(a+b)


a = int(input("enter first number: "))
b = int(input("enter second number: "))
c = int(input("enter the third number: "))
print("average of three numbers: ",cal_avg(a,b,c))
print("sum of first two numbers: ", cal_sum(a,b))
print("sum of last two numbers: ", cal_sum(b,c))#here b,c are argument not parameter
```

```python
cities = ["ranchi", "jodhpur","purnea","surat"]
numbers = [1,2,3,4,5]

def print_len(list):
  print("length of lists:", list)
  print(len(list))

def print_list(list):
  for item in list:
    print(item)


print_len(cities)
print_len(numbers)

print_list(cities)
print_list(numbers)
```

```python
n = int(input("enter the number you want factorial: "))

def factorial_n(n):
  fact = 1
  for i in range(1, n+1):
    fact *= i
    print(fact)


factorial_n(n)
```
