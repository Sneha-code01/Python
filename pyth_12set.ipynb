```python
#sets is collection of the unordered items
#each element in set must be unque and immutable
#lists and dictionary are mutable values and cant be stored in sets
collection = {1,2,3,4,1,2,2,2,"shruti"}
#sets do not print repeated value
print(collection)
print(type(collection))
print(len(collection))
```

```python
group = set()
print(type(group))
#sytx for empty sets
```

```python
#set method
group.add(4)
print(group)
collection = {1,2,3,4,5,6,7}
collection.remove(1)
print(collection)
collection.pop()
collection.union(group)
collection.intersection(group)
```
