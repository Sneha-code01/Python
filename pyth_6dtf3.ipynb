```python
import numpy as np
import pandas as pd
#functions and methods
#1. value_count(series+dtf) in series countss the frequency of ddata appering in series and in dtf it counts no of time a row is repeated
diabtes = pd.read_csv("/content/diabetes.csv")
print(diabtes.value_counts())
```

```python
ipl = pd.read_csv("/content/ipl-matches.csv")
#take player who won most of the man of the match award of final/quater /semi
spl_mth = ipl[~ipl['MatchNumber'].str.isdigit()]
print(spl_mth[['Player_of_Match']].value_counts())
```

```python
#ques says after winning toss team do bating or balling/
print(ipl['TossDecision'].value_counts().plot(kind='pie'))
```

```python
#total no of matches played by teams
print(ipl['Team1'].value_counts() + ipl['Team2'].value_counts().sort_values(ascending=False))
```

```python
#sort_values('name',na_position='first',ascending=False) nan is by efault set at last
#rank
import numpy as np
batsman = pd.read_csv("/content/batsman_runs_ipl.csv")
print(batsman)
run = batsman['batsman_run'].rank(ascending=False)
batsman['batter_rank']= run
print(batsman.sort_values('batter_rank'))
```

```python
#sort index sort the data according to index
#set_index set the index as per your prefrence from any column
#reset_index undo set index opt
#reoplce existing column without losing
print(batsman.set_index('batter').reset_index('batter').set_index('batter_rank'))
#reset index makes series tp dtf
```
