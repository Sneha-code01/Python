```python
info = {
    "key" : "value", "name": "Sneha", "learning": "python", "age" : 19, "adult": True,
}
print(info)
print(type(info))
info["name"] = "Shruti"
print(info)

#dictonary are unoredered and mutable and dont allow duplicate keys
```

```python
student = {
    "name" : "Sneha Shruti",
    "Subject" : {
        "phy" : 98,
        "chem" : 90,
        "mth" : 97,
    }
}
print(student)
print(student["Subject"]["phy"])
#nested dictionary
```

```python
#dictionary methods
print(list(student.values()))
#will print the values of the items
print(list(student.keys()))
print(list(student.items()))
#lists the dic in pair as tuples
print(list(student.get("name")))
#get return the key accordinng to value ie if the value of key is not present than return none as value
print(student["name"])
#it return if only value is present else shows error
student.update({"city": "ranchi"})
print(student)
```
