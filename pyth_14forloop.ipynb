```python
#for loops are used for sequential traversal
#for travelling lists string and tuples
```

```python
nums = [1,2,3,4,5]

for val in nums:
  print(val)
```

```python
tup = (1,4,9,16,25,36,49,64,81,100)
x = int(input("enter the number you wanna find: "))

idx = 0
for el in tup:
  if(el == x):
    print("number found at index : ", idx)
  idx += 1
```

```python
#range function return a sequence of number starting from zero by default
for el in range(1,5):
  print(el)
```

```python
for el in range(1,15,3):#start,stop,step size
  print(el)
```
